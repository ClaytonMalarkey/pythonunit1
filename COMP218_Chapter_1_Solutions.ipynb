## Exercise 1.1 — Number-system conversions

Convert each value to binary. The decimal, hexadecimal, and octal examples are converted by repeatedly dividing by 2 or by expanding each digit into its binary equivalent.

```python
# Decimal to binary
values_decimal = [78, 121, 29]
for value in values_decimal:
    print(value, "=", bin(value)[2:])

# Hexadecimal to binary: each hex digit represents four bits
hex_values = ["1F", "3E", "CD"]
for value in hex_values:
    print(value, "=", format(int(value, 16), "b"))

# Octal to binary: each octal digit represents three bits
octal_values = ["27", "33", "52"]
for value in octal_values:
    print(value, "=", format(int(value, 8), "b"))
```

**Explanation / expected result**

Answers:
(78)10 = 1001110₂; (1F)16 = 11111₂; (27)8 = 10111₂
(121)10 = 1111001₂; (3E)16 = 111110₂; (33)8 = 11011₂
(29)10 = 11101₂; (CD)16 = 11001101₂; (52)8 = 101010₂

The code uses Python's int(value, base) conversion and bin()/format() to check the manual conversions.

## Exercise 1.2 — Binary arithmetic

Work each operation in binary. Python's int(..., 2) can verify the decimal value of each result.

```python
operations = [
    ("10111", "+", "1101"),
    ("1101010", "-", "101101"),
    ("10101", "+", "1110"),
    ("10101", "-", "1111"),
    ("1101", "*", "101"),
    ("111011", "/", "11"),
]
for left, operator, right in operations:
    a, b = int(left, 2), int(right, 2)
    if operator == "+": result = a + b
    elif operator == "-": result = a - b
    elif operator == "*": result = a * b
    else: result = a / b
    print(f"{left} {operator} {right} = {result} (decimal)")
    if result == int(result):
        print("Binary result:", bin(int(result))[2:])
```

**Explanation / expected result**

10111 + 1101 = 100100₂ (36)
1101010 − 101101 = 111101₂ (61)
10101 + 1110 = 100011₂ (35)
10101 − 1111 = 110₂ (6)
1101 × 101 = 1000001₂ (65)
111011 ÷ 11 = 10011₂ (19)

The last division is exact: 59 ÷ 3 = 19 remainder 2? In binary, 111011₂ is 59 and 11₂ is 3, so the quotient is 19 remainder 2. The code's floating result is 19.666...; the integer quotient is 10011₂ with remainder 10₂.

## Exercise 1.4 — Programming-language features

The following descriptions compare the language categories named in the exercise.

| Type | Main feature | Example / note |
|---|---|---|
| Machine code | Binary instructions executed directly by a processor | Fast for the target CPU but difficult for people to read. |
| Assembly language | Human-readable mnemonics mapped closely to machine instructions | Requires an assembler; hardware-specific. |
| High-level language | More human-friendly abstractions | Python, C, Java; translated or interpreted for execution. |
| Procedural | Organizes a program as procedures/functions and ordered steps | C and many Python programs. |
| Structured | Uses clear control structures such as sequence, selection, and iteration | Avoids unstructured jumps; supported by Python. |
| Imperative | Describes commands that change program state | Assignment, loops, and conditionals. |
| Functional | Emphasizes expressions and function evaluation, often minimizing mutable state | Lisp; Python also supports functional techniques. |
| Logic | Expresses facts and rules; computation follows logical inference | Prolog. |
| Object-oriented | Organizes data and behaviour into objects/classes | Python, Java, C++. |

**Explanation / expected result**

Categories can overlap. Python is high-level and multi-paradigm: it supports procedural, imperative, functional, and object-oriented styles. Structured programming is a way of organizing control flow rather than a separate hardware-level language.

## Project 2 — History of programming languages

A short overview of how programming languages developed and why different styles exist.

Early computers were programmed with machine-language instructions, represented as binary patterns that the processor could execute directly. Assembly language made these instructions easier for people to read by using mnemonics, but it remained closely tied to a particular processor and required an assembler. High-level languages made programs more expressive and portable by allowing programmers to describe operations in a more human-friendly form.

The textbook discusses Konrad Zuse's Plankalkül design from 1942–1945, ALGOL 58 and ALGOL 60, and languages such as COBOL, Fortran, Ada, Lisp, and Prolog. These languages reflect different purposes and programming approaches. Procedural and imperative languages describe a sequence of actions. Structured programming uses clear sequence, selection, and iteration. Functional programming emphasizes functions and expression evaluation. Logic programming describes facts and rules. Object-oriented programming groups data and related operations into objects.

High-level languages have trade-offs. They are generally easier to read, maintain, and develop with, but they need translation or an interpreter/runtime and may offer less direct control over hardware than assembly. Machine code is directly executable but difficult to write and maintain. Assembly offers more control but is processor-specific. Python is a popular high-level language because it is readable and supports multiple programming styles.
